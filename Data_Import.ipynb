```python
import pandas as pd
from urllib.parse import quote_plus
from sqlalchemy import create_engine


path = "C:/Users/Kbelw/Data analyst (powerbi + sql)/Project/Dataset.xlsx"
df = pd.read_excel(path)

username = "root"
password = "Krishnakrish26@1978"
host = "localhost"
port = 3306
database = "Sales_Performance_And_Profitability_Analytics"

encoded_password = quote_plus(password)

connection_string = (
    f"mysql+pymysql://{username}:{encoded_password}@{host}:{port}/{database}"
)

engine = create_engine(connection_string)

# create different dataframew from excel sheet

Orders_df = df["Orders"]
Returns_df = df["Returns"]
People_df = df["People"]

# Data load to mysql

Orders_df.to_sql(
    name="orders",
    con=engine,
    if_exists="replace",
    index=False
)
Returns_df.to_sql(
    name="returns",
    con=engine,
    if_exists="replace",
    index=False
)
People_df.to_sql(
    name="people",
    con=engine,
    if_exists="replace",
    index=False
)
print("successfully loaded")
```
